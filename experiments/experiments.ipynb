###  Get all movies

```python
import requests
import os 
from dotenv import load_dotenv
import pandas as pd
load_dotenv()
API_KEY = os.getenv('DB_API_TOKEN')
url = 'https://api.themoviedb.org/3/discover/movie'
headers = {
    "accept": "application/json",
    "Authorization": f"Bearer {API_KEY}",

}
# first get request to get total pages and results 
response = requests.get(url, headers=headers)
results = response.json()
total_pages = results['total_pages']
total_results = results['total_results']
#initialize results list
results_list = []
#iterate over all pages
for page_number in range(1,total_pages): 
    if page_number%50 == 0:
        print(page_number) 
    search_url = f'https://api.themoviedb.org/3/discover/movie?page={page_number}'
    response = requests.get(url, headers=headers)
    results = response.json()
    results_list.append(results['results'])


results_dataframe = pd.DataFrame(results_list)
results_dataframe.to_csv("Movies_export.csv",index = False)
    
```
